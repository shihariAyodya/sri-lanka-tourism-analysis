```python
df = pd.read_excel(file_2025, header=2)
df.head()
```

```python
df = df.dropna(axis=1, how="all")
df.columns = df.columns.str.strip()
df.columns
```

```python
df = df.dropna(subset=["Country"])
df.shape
```

```python
df = df[df["Country"] != "Country"]
df = df.reset_index(drop=True)
df.shape
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

df["Rank"] = df["Rank"].astype(int)
df[months + ["Total"]] = df[months + ["Total"]].astype(int)
df.dtypes
```

```python
df.isna().sum()
```

```python
check = df[months].sum(axis=1)
df[check != df["Total"]]
```

```python
df = df.rename(columns={
    "Rank": "No",
    "Jan": "January",
    "Feb": "February",
    "Mar": "March",
    "Apr": "April",
    "Jun": "June",
    "Jul": "July",
    "Aug": "August",
    "Sep": "September",
    "Oct": "October",
    "Nov": "November",
    "Dec": "December"
})
df["Country"] = df["Country"].str.strip()
df["Year"] = 2025
df.to_csv("data/clean/clean_2025.csv", index=False)
os.listdir("data/clean")
```
