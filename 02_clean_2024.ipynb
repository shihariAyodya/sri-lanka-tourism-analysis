```python
df = pd.read_excel(file_2024, header=1)
df.head()
```

```python
df = df.dropna(axis=1, how="all")
df.columns = df.columns.str.strip()
df.columns
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df = df.dropna(subset=["Country"])
df.shape
```

```python
months = ["Jan", "Feb", "Mar", "April", "May", "June",
          "July", "August", "Sep", "Oct", "Nov", "Dec"]

check = df[months].sum(axis=1)
df[check != df["Total"]]
```

```python
df = df.rename(columns={
    "Jan": "January",
    "Feb": "February",
    "Mar": "March",
    "April": "April",
    "Sep": "September",
    "Oct": "October",
    "Nov": "November",
    "Dec": "December",
    "June": "June",
    "July": "July",
    "August": "August"
})
df.columns
```

```python
df["No"] = df["No"].astype(int)
df["Country"] = df["Country"].str.strip()
df["Year"] = 2024
df.head()
```

```python
df.to_csv("data/clean/clean_2024.csv", index=False)
os.listdir("data/clean")
```
